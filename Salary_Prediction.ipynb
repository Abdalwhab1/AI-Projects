```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

data= {
    'Experience':[1, 3, 6, 10],

    'Salary': [3000, 4200, 7000, 9200]
}

df = pd.DataFrame(data)
print("--- Employees Dataset ---")
print(df)

X = df[['Experience']]
Y = df['Salary']

model = LinearRegression()
model.fit(X, Y)

predicted_salary = model.predict([[8]])
print(f"\nPredicted Salary for 8 years of experience: {predicted_salary[0]:.0f}")

plt.scatter(X,Y, color = 'blue', label = 'Actual Data')
plt.plot(X, model.predict(X), color = 'green', label= 'AI Prediction Line')
plt.title("Salary Prediction Based on Experience", color = 'red')
plt.xlabel("Years of Experience", color = 'red')
plt.ylabel("Salary", color = 'red')
plt.legend()
plt.grid(True)
plt.show()
```
